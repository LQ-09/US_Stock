```python
unique_tickers = df.index.get_level_values('Ticker').unique()
```

```python
save_directory = r'D:\Project\US_Stock\data\daily_folder'  # Save to local file.
os.makedirs(save_directory, exist_ok=True)  # if file does not existsl, create
```

```python
for ticker in unique_tickers:
    # filter out each ticker
    df_ticker = df.loc[df.index.get_level_values('Ticker') == ticker]
    
    # define the path to save
    parquet_file_path = os.path.join(save_directory, f'{ticker}.parquet')
    
    # save as parquet file
    df_ticker.to_parquet(parquet_file_path, index=True)

print("Data saved successfully for each stock in parquet！")
```
